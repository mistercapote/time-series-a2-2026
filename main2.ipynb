## Modelagem SARIMA

### Diferenciação das séries

Como os dados possuem sazonalidade conhecida de 7 dias, foi analisado o efeito de uma diferenciação sazonal com período $m=7$:

```python
for label in labels:
    # Aplica diferenciação sazonal e não sazonal
    diff_series = train_data[label].diff(7).dropna()
    diff_dates = train_data["date"].loc[diff_series.index]

    viz.grafico_temporal(
            diff_series,
            diff_dates,
            titulo=f"{label} ao longo do tempo (Conjunto de Treino) Diferenciada (d=0, D=1, m=7)"
        )

    viz.grafico_ACF(
        diff_series,
        diff_dates,
        titulo=f"ACF Diferenciada (d=0, D=1, m=7) - {label}",
        nlags=22,
    )

    viz.grafico_PACF(
        diff_series,
        diff_dates,
        titulo=f"PACF Diferenciada (d=0, D=1, m=7) - {label}",
        nlags=22,
    )
```

Além disso, para controlar o efeito de tendência, foi escolhido aplicar uma diferenciação não sazonal em cada uma das séries.

A partir das análises, foi decidido aplicar uma diferenciação sazonal com período $m = 7$ e uma diferenciação não sazonal em cada uma das séries, para modelar a tendência de crescimento em cada série e a tendência sazonal.

```python
for label in labels:
    # Aplica diferenciação sazonal e não sazonal
    diff_series = train_data[label].diff(7).diff(1).dropna()
    diff_dates = train_data["date"].loc[diff_series.index]

    viz.grafico_temporal(
                diff_series,
                diff_dates,
                titulo=f"{label} ao longo do tempo (Conjunto de Treino) Diferenciada (d=1, D=1, m=7)"
            )

    viz.grafico_ACF(
        diff_series,
        diff_dates,
        titulo=f"ACF Diferenciada (d=1, D=1, m=7) - {label}",
        nlags=22,
    )

    viz.grafico_PACF(
        diff_series,
        diff_dates,
        titulo=f"PACF Diferenciada (d=1, D=1, m=7) - {label}",
        nlags=22,
    )
```

### Escolha dos parâmetros do SARIMA

Nas três séries, é possível perceber um corte na ACF no lag $1$ e um corte sazonal no lag $7$, com decaimento não sazonal e sazonal na PACF. Portanto, optou-se, inicialmente por um $SARIMA(0,1,1)(0,1,1)_7$ para as três séries, resultando nos seguintes resultados para ACF e PACF dos resíduos:

```python
for label in labels:
    resid = models.sarima(train_data[label], (0, 1, 1), (0, 1, 1, 7), resid=True)

    viz.grafico_ACF(
        resid,
        train_data["date"].loc[resid.index],
        titulo=f"ACF SARIMA (0,1,1) (0,1,1,7) - {label}",
        nlags=22,
    )

    viz.grafico_PACF(
        resid,
        train_data["date"].loc[resid.index],
        titulo=f"PACF Diferenciada (0,1,1) (0,1,1,7) - {label}",
        nlags=22,
    )
```

### Proposta de mudança nos parâmetros

Analisando os resíduos, é possível perceber, nas três séries, que ACF e PACF cortam no lag $1$, evidenciando que ainda existe dependência temporal nos resíduos. Por isso, como o corte foi no primeiro lag em ambos os gráficos, optou-se por adicionar uma unidade no $p$ e comparar com adicionar uma unidade no $q$, utilizando AIC e BIC:

```python
modelos = [
    ((0, 1, 2), (0, 1, 1, 7)),
    ((1, 1, 1), (0, 1, 1, 7))
]

tabela = model_checking.comparar_sarima(modelos, train_data, labels=labels)

print("=== COMPARAÇÃO DE CRITÉRIOS DE INFORMAÇÃO (AIC/BIC) NO TREINO ===")
display(tabela)
```

Avaliando exclusivamente os critérios de informação calculados no conjunto de treino (com as restrições de estacionariedade e invertibilidade ativas):

- Na série _store\_total_, o modelo $SARIMA(0,1,2)(0,1,1)_7$ apresentou os menores valores de AIC (`28191,4` vs. `28199,1`) e BIC (`28213,5` vs. `28221,3`), sendo portanto selecionado.

## Parte 2 — Previsão com covariáveis

### Preparação das covariáveis e escolha dos parâmetros do SARIMAX

Cada série (`store_total`, `FOODS` e `HOBBIES`) possui ajuste independente em sua própria tabela (o comportamento semanal e o nível de preço de uma categoria não se misturam com os das demais). A partir de `dados/covariaveis.csv` e das datas conhecidas na origem (`2016-03-27`), entraram no modelo de cada série **46 covariáveis**:
- **Programa SNAP (`snap_CA`):** indicadora binária (0/1) dos dias em que há benefício SNAP na Califórnia;
- **Preço da própria série (`preco_store_total`, `preco_FOODS` ou `preco_HOBBIES`):** padronizado utilizando **exclusivamente a média e o desvio-padrão do treino** (até `2016-03-27`), descartando os preços das outras duas séries;
- **Calendário numérico:** `semana_do_ano`, `mes` e `ano`;
- **Dia da semana:** 7 indicadoras binárias (`dia_da_semana_0` a `dia_da_semana_6`);
- **Eventos e feriados:** 4 indicadoras de tipo de evento (`event_type_Cultural`, `National`, `Religious`, `Sporting`, unificando `event_type_1` e `event_type_2`) e 30 indicadoras de nome de evento (`event_name_*`, unificando `event_name_1` e `event_name_2`).

Com essas covariáveis preparadas, reavaliamos a escolha das ordens $(p,d,q)\times(P,D,Q)_7$ para o **SARIMAX** (com restrições estritas de estacionariedade e invertibilidade) em cada uma das três séries. Partindo de $SARIMAX(0,1,1)(0,1,1)_7$, analisamos a ACF e a PACF dos resíduos no conjunto de treino:

```python
tabela_treino_task2, tabela_validacao_task2 = features.montar_treino_e_validacao()

for label in labels:
    resid_sarimax = models.sarimax(
        label, (0, 1, 1), (0, 1, 1, 7), tabela_treino=tabela_treino_task2, tabela_validacao=tabela_validacao_task2, resid=True
    )

    viz.grafico_ACF(
        resid_sarimax,
        train_data["date"].loc[resid_sarimax.index],
        titulo=f"ACF SARIMAX (0,1,1) (0,1,1,7) - {label}",
        nlags=22,
    )

    viz.grafico_PACF(
        resid_sarimax,
        train_data["date"].loc[resid_sarimax.index],
        titulo=f"PACF SARIMAX (0,1,1) (0,1,1,7) - {label}",
        nlags=22,
    )
```

Como os resíduos de $SARIMAX(0,1,1)(0,1,1)_7$ ainda exibem autocorrelação significativa no primeiro lag nas três séries, comparamos as ordens candidatas $(0,1,1)\times(0,1,1)_7$, $(0,1,2)\times(0,1,1)_7$ e $(1,1,1)\times(0,1,1)_7$ por meio dos critérios de informação AIC e BIC no conjunto de treino:

```python
ordens_candidatas_sarimax = [
    ((0, 1, 1), (0, 1, 1, 7)),
    ((0, 1, 2), (0, 1, 1, 7)),
    ((1, 1, 1), (0, 1, 1, 7)),
]

tabela_sarimax = model_checking.comparar_sarimax(ordens_candidatas_sarimax, tabela_treino_task2, labels)
display(tabela_sarimax)
```

Com a introdução das covariáveis exógenas e mantendo as restrições de estacionariedade e invertibilidade no filtro de Kalman (`mle_regression=False`), a seleção pelo menor AIC e BIC no conjunto de treino define as seguintes ordens por série:

Verificando a ACF e a PACF dos resíduos dos modelos `SARIMAX_DICT` escolhidos:

### Análise comparativa entre `linear`, `sarimax` (SARIMA com covariáveis), `sarima` e `naive_sazonal`

Na mesma janela de validação de 28 dias (`2016-03-28` a `2016-04-24`), utilizando exatamente o mesmo conjunto de covariáveis no `linear` e no `sarimax`:

- **`store_total`:** O `sarimax` $(0,1,2)\times(0,1,1)_7$ alcança o menor erro da série nas três métricas (MAE `223,615`, RMSE `270,515` e MASE `0,514`), seguido pelo `linear` (`lasso`, $\alpha=25$, recursivo, com MAE `249,609`, RMSE `313,954` e MASE `0,574`). Ambos superam tanto o `sarima` sem covariáveis $(0,1,2)\times(0,1,1)_7$ (MAE `264,280`, RMSE `341,163` e MASE `0,608`) quanto o `naive_sazonal` (MAE `459,179`, RMSE `655,467` e MASE `1,056`).
- **`FOODS`:** O `sarimax` $(1,1,1)\times(0,1,1)_7$ também obtém o menor erro nas três métricas (MAE `175,090`, RMSE `212,846` e MASE `0,512`), superando com folga o `sarima` sem covariáveis $(1,1,1)\times(0,1,1)_7$ (MAE `209,347`, RMSE `254,702` e MASE `0,612`), o `linear` (`lasso`, $\alpha=50$, recursivo, com MAE `215,084`, RMSE `267,967` e MASE `0,629`) e o `naive_sazonal` (MAE `343,857`, RMSE `420,116` e MASE `1,005`).
- **`HOBBIES`:** Tanto o `linear` (`ridge`, $\alpha=500$, recursivo, com MAE `76,168`, RMSE `91,211` e MASE `0,834`) quanto o `sarimax` $(1,1,1)\times(0,1,1)_7$ (MAE `77,488`, RMSE `89,576` e MASE `0,849`) superam com ampla vantagem o `naive_sazonal` nas três métricas (MAE `102,250`, RMSE `146,653` e MASE `1,120`). Em comparação ao `sarima` sem covariáveis $(1,1,1)\times(0,1,1)_7$ (MAE `73,732`, RMSE `94,593` e MASE `0,808`), o `sarimax` e o `linear` reduzem o RMSE (de `94,593` para `89,576` e `91,211`, respectivamente), enquanto o `sarima` mantém ligeira vantagem em MAE e MASE.

Em síntese, nem o `linear` nem o `sarimax` perderam para o `naive_sazonal` em nenhuma das três séries ou métricas (todos com MASE bem abaixo de 1), e a inclusão das covariáveis no `sarimax` reduziu substancialmente os erros do `sarima` em `store_total` e `FOODS`, além de melhorar o RMSE em `HOBBIES`.
